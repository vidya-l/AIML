###1. Simple ATM Calculator


```python
#Write a program that takes the user's account balance and withdrawal amount.
#If sufficient balance is available, deduct the amount and display the updated balance.
#Otherwise, print an error message.

# Get user input
balance = eval(input("Enter your account balance: "))
withdraw = eval(input("Enter withdrawal amount: "))

# Check balance
if withdraw <= balance:
    #calculate balance and print
    balance = balance - withdraw
    print("Transaction successful. Remaining balance", balance)
else:
    #insufficient balance error
    print("Error: Insufficient balance.")
```

###2. Electricity Bill Generator

```python
#Create a bill calculator where the user enters the number of units consumed. The rate is
#₹5 per unit for the first 100 units, ₹7 for the next 100, and ₹10 for anything above 200
#units. Calculate and display the total bill.
def calculate_bill_amount():
    #get user input
    units = int(input("Enter units consumed: "))
    bill = 0
    #check if units less than or equal to 100, then units * 5
    if units <= 100:
        bill = units * 5
    #check if units less than or equal to 200, then units * 5
    elif units <= 200:
        bill = (100 * 5) + (units - 100) * 7
    #for anyting greater 200
    else:
        bill = (100 * 5) + (100 * 7) + (units - 200) * 10
    print("Bill amount is", bill)
calculate_bill_amount()
```

##4. Mobile Recharge Validity


```python
# Write a program that takes the recharge date and today’s date as input and checks
# whether the 28-day recharge plan is still valid or expired.
try:
  #importing datetime module
  from datetime import datetime
  #accepting recharge date
  recharge_date = input("Enter recharge date")
  #convert recharge date string to date
  recharge = datetime.strptime(recharge_date, "%Y-%m-%d")
  #getting todays date
  today = datetime.today();
  #get date difference
  days = (today - recharge).days
  if days >= 28:
      print('Plan expired. Recharge again.')
  else:
      print('Plan valid.')
except ValueError:
    #throw date format error
    print("Invalid date format! Please use YYYY-MM-DD")
```

##6. Sum of even numbers


```python
# Write a program that takes a number as input and prints the sum of all even numbers
#from 1 to that number.
number = int(input('Enter number'))
total = 0
for i in range (0,number+1, 2):
    total += i
print('Sum of even numbers from 1 to ',number, 'is', total)
```

##7. Dictionary – Student Marks Summary

```python
# Write a program that takes input of 3 students’ names and their marks, stores them in a
# dictionary, and prints the average marks.

#student limit variable
limit  = 3
#declaring empty list
students = {}
for i in range(1,limit+1):
    #enter name
    name = (input("Enter name"))
    #enter mark
    mark = (int(input("Enter marks")))
    #assign to dictionary
    students[name] = mark

#calculate average
average_marks = sum(students.values()) / len(students)
print('Average Marks', average_marks)
```

##8. Dictionary – Word Frequency

```python
from re import split
# Write a program that takes a sentence and displays the frequency of each word using a
#dictionary.

sentence = input("Enter sentence")
words = sentence.split()
word_dict = {}
for i in words:
    if i in word_dict:
        word_dict[i] += 1
    else:
        word_dict[i] = 1

print(word_dict)
```

##9. Functions – Check Prime

```python
#Write a function that checks whether a number is prime. Call the function with user input.
#get number as input
number = int(input("Enter a number"))
#set prime as true
prime = True;
#if number is less than or equal to 1 then false
if number <= 1:
    prime = False
else:
    #check number is divisible by each number
    for i in range(2,number):
        #if divisible then not a prime
        if number % i == 0:
            prime = False;
if prime:
    print(number, 'is a prime number');
else:
    print(number, 'is not a prime number');
```

##10. Vowel Counter


```python
#Write a function that takes a string and returns the number of vowels in it.
#get number as input
string = input("Enter a string")
#vowels
vowels = ['a', 'e', 'i', 'o', 'u']
count = 0
for i in string:
    if i in vowels:
        count += 1
print("Number of vowels", count)
```

## 11. Count Divisible Numbers

```python

# Write a program that takes two numbers as input: a range limit N and a divisor D. Count
# how many numbers between 1 and N are divisible by D.
limit = int(input("Enter the range limit (N)"))
divisor = int(input("Enter the divisor (D)"))
count = 0
for i in range(1, limit + 1):
    if i % divisor == 0:
        count += 1
print('Count of numbers divisible by', divisor, ':', count)
```
